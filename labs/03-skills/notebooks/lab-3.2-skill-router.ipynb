```python
import sys
from pathlib import Path

LAB = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(LAB))

from src.router import SkillRouter
from src.registry import build_default_registry
from src.schemas import RouteRequest

router = SkillRouter(build_default_registry())
```

```python
samples = [
    RouteRequest(text="Validate the JSON syntax only", has_dfd=True),
    RouteRequest(text="Does a path from an external entity cross a trust boundary?", has_dfd=True),
    RouteRequest(text="Which SEC-DFD controls apply to data stores?", has_dfd=False),
    RouteRequest(text="Run a full compliance evaluation", has_dfd=True),
    RouteRequest(text="", has_dfd=True),
]

print(f"{'Route':12} {'#tools':>6} {'≈tokens':>8}  skills")
print("-" * 70)
full_tokens = router.registry.estimate_tool_tokens()
for req in samples:
    d = router.decide(req)
    toks = router.registry.estimate_tool_tokens(only=d.skill_names)
    print(f"{d.route.value:12} {len(d.skill_names):6d} {toks:8d}  {d.skill_names}")

print(f"\nFull tool set tokens: {full_tokens}")
```

## Routing rules (document for submission)

- `syntax` – validation / schema-only language, no path or policy terms
- `structural` – path / trust-boundary / external-entity language with a DFD
- `policy` – control / baseline questions without a DFD payload
- `full` – explicit evaluation language, or bare DFD upload
